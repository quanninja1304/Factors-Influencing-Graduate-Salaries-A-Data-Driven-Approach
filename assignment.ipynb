```python
import pandas as pd
import statsmodels.api as sm

# Load the cleaned dataset (adjust path if needed)
file_path = 'cleaned_dat_2.xlsx'
df = pd.read_excel(file_path, sheet_name=0)

# Define dependent and independent variables
y = df['Starting_Salary']
X = df.drop(columns=['Starting_Salary'])

# Add intercept to the model
X = sm.add_constant(X)

# Fit the OLS regression model
model = sm.OLS(y, X).fit()

# Display regression results
print(model.summary())

# Optional: Save coefficients and standard errors to a DataFrame
results_df = pd.DataFrame({
    'Coefficient': model.params,
    'Std_Error': model.bse
})

# Optional: Add fitted values to your dataframe
df['Fitted_Starting_Salary'] = model.fittedvalues

# Preview a few rows
print(df[['Starting_Salary', 'Fitted_Starting_Salary']].head())
```
